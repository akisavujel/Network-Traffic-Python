```python
# IMPORT REQUIRED LIBRARIES IN NOTEBOOOK

import pandas as pd
```

```python
# CREATE DATAFRAME USING DATASET OF  CICIDS (Canadian Institute for Cybersecurity Intrusion Detection System) NETWORK TRAFFIC DATASET

df = pd.read_csv('/kaggle/input/datasets/akisavujel/network-traffic/ids.csv')
```

```python
# REQUIRED COLUMNS FOR ANALYSIS

columns = ['Destination Port', 'Flow Duration', 'Total Fwd Packets', 
           'Total Backward Packets', 'Total Length of Fwd Packets', 
           'Total Length of Bwd Packets', 'Fwd Packet Length Mean', 
           'Bwd Packet Length Mean', 'Flow Bytes/s', 'Flow Packets/s', 
           'Packet Length Mean', 'Packet Length Std', 'Average Packet Size', 
           'Active Mean', 'Idle Mean', 'Label']
```

```python
# CREATE NEW DATAFRAME

df_filtered = df[columns]
```

```python
# LAST 50 VALUES OF NEW DATASET

df_filtered.tail(50).style\
  .background_gradient(cmap='Purples')\
  .highlight_null(color='red')\
  .set_caption('📋 Filtered Dataset')\
  .set_table_styles([{
      'selector': 'caption',
      'props': [('font-size', '14px'),
                ('font-weight', 'bold'),
                ('color', '#2d5a4f')]
  }])
```

```python
# TOTAL ROWS AND COLUMN IN NEW DATASET

print("=" * 60)
print("   ✅ FILTERED DATAFRAME CREATED SUCCESSFULLY!")
print("=" * 60)
print("   Total Rows    :", df_filtered.shape[0])
print("   Total Columns :", df_filtered.shape[1])
print("=" * 60)
```

```python
# FIRST THREE VALUES OF DATASET

print("=" * 60)
print("   📋 TOP 3 VALUES OF FILTERED DATASET")
print("=" * 60)
print("\n")
df_filtered.head(3).style.background_gradient(cmap='Greens')
```

```python
# SHAPE OF NEW DATASET

print("=" * 50)
print("   📐 SHAPE OF NEW DATAFRAME")
print("=" * 50)
print(df_filtered.shape)
```

```python
# MISSING VALUES IN ROWS

print("=" * 50)
print("   🔍 MISSING VALUES PER COLUMN")
print("=" * 50)
print(df_filtered.isnull().sum())
```

```python
# TOTAL SUM OF MISSING VALUES

print("=" * 50)
print("   🔍 TOTAL MISSING VALUES")
print("=" * 50)
print(df_filtered.isnull().sum().sum())
```

```python
# SHAPE OF DATASET AFTER DELETING EMPTY ROWS

df_filtered = df_filtered.dropna()

print("=" * 50)
print("   ✅ SHAPE AFTER REMOVING MISSING VALUES")
print("=" * 50)
print(df_filtered.shape)
```

```python
# COLUMNS SEECTED FOR ANALYZING RELATIONS

cols = ['Flow Duration', 'Packet Length Mean']
```

```python
# CALCULATE OF FLOW DURATION AND PACKET LENGTH MEAN

sum_result = df_filtered[cols].sum()
mean_result = df_filtered[cols].mean()
std_result = df_filtered[cols].std()
skew_result = df_filtered[cols].skew()
kurt_result = df_filtered[cols].kurt()
```

```python
print("=" * 50)
print("   ➕ SUM")
print("=" * 50)
print(sum_result)

print("\n")

print("=" * 50)
print("   📊 MEAN")
print("=" * 50)
print(mean_result)

print("\n")

print("=" * 50)
print("   📉 STANDARD DEVIATION")
print("=" * 50)
print(std_result)

print("\n")

print("=" * 50)
print("   📐 SKEWNESS")
print("=" * 50)
print(skew_result)

print("\n")

print("=" * 50)
print("   📈 KURTOSIS")
print("=" * 50)
print(kurt_result)
```

```python
# THE COLUMN LABEL CONTAINS TEXT VARIABLE WHICH DOESNOT WORK WITH CORR

correlation_matrix = df_filtered.drop(columns=['Label']).corr()
```

```python
# CORRELATION MATRIX OF FLOW DURATION AND PACKET LENGTH MEAN

print("=" * 50)
print("   🔗 CORRELATION MATRIX")
print("=" * 50)
correlation_matrix.style.background_gradient(cmap='Greens')
```

```python
# CONVERT THE CORR TABLE TO LIST

corr_pairs = correlation_matrix.unstack() 

# REMOVE WHERE CORR IS PERFECT 1

corr_pairs = corr_pairs[corr_pairs != 1] 

# DROP REPEATED PAIRS 

corr_pairs = corr_pairs.drop_duplicates() 
corr_pairs_sorted = corr_pairs.abs().sort_values(ascending=False) 
```

```python
# TOP FIVE CORRELATION FEATURE IN DATAFRAME

print("=" * 50)
print("   🔝 TOP 5 CORRELATED FEATURES")
print("=" * 50)
print(corr_pairs_sorted.head(5).to_string())
```

```python
# HYPOTESTICAL TESTING TO VALIDATE OUR DATA

from scipy import stats
```

```python
# MAKE VARIABLE TO STORE FILTERED ROWS

normal = df_filtered[df_filtered['Label'] == 'Normal Traffic']['Flow Duration']
attack = df_filtered[df_filtered['Label'] == 'Attack']['Flow Duration']
```

```python
# USE T TEST AND P VALUE FOR TESTING

t_stat, p_value = stats.ttest_ind(normal, attack)
```

```python
# T STAT VALUE AND P VALUE

print("=" * 50)
print("   📊 T-STATISTIC")
print("=" * 50)
print(t_stat)

print("\n")

print("=" * 50)
print("   📉 P-VALUE")
print("=" * 50)
print(p_value)
```

```python
# CHECK IF NULL IS ACCEPTED OR REJECTED

print("=" * 50)
print("   🔬 HYPOTHESIS TEST RESULT")
print("=" * 50)
if p_value < 0.05:
    print("   ✅ Result: Reject Null Hypothesis")
    print("   There is a significant difference")
else:
    print("   ❌ Result: Accept Null Hypothesis")
    print("   There is no significant difference")
```
